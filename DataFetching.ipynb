```python
from pyspark.sql.functions import col
statusinprocesdf = servicesdf.filter(col("Status") == 'In-Process')
display(statusinprocesdf)
```

```python
from pyspark.sql.functions import col
completeddf = servicesdf.filter(col("Status") == 'Completed')
display(statusinprocesdf)
```

```python
%sql
drop table statuscompleted;
drop table statusinprocess;
```

```python

completeddf.write.mode("overWrite").saveAsTable("training8am.statuscompleted")
statusinprocesdf.write.mode("overwrite").saveAsTable("training8am.statusinprocess")
```

```python
%sql
create view if not exists training8am.statusimprocessview as
select * from training8am.statusinprocess
```
