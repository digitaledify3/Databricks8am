> **STORED PROCEDURES**

```python
%sql
CREATE PROCEDURE IF NOT EXISTS procedurename()
SQL SECURITY INVOKER/DEFINER
LANGUAGE SQL
AS
BEGIN
    SQL STATEMENTS
END;
```

- **Security Invoker** - Procedure runs with the permissions of the person calling it.
- **Security Definer** - Anyone having the execute grant/permission can run this store proc.

```python
%sql
CREATE PROCEDURE IF NOT EXISTS GetCompletedServices()
SQL SECURITY INVOKER
LANGUAGE SQL
AS
BEGIN
    SELECT * 
    FROM training8am.services 
    where status = 'Completed';
END;
```

```python
%sql
CALL GetCompletedServices();
```

```python
%sql
CREATE PROCEDURE IF NOT EXISTS training8am.customerservicehistory()
SQL SECURITY INVOKER
LANGUAGE SQL
AS
BEGIN
    SELECT
    c.CustomerName,
    s.ServiceID,
    s.Status,
    s.servicecharges
FROM training8am.customerview c
left join training8am.services s
on c.CustomerId = s.CustomerId;
END
```

```python
%sql
call training8am.customerservicehistory()
```

```python
%sql
-- ServiceDetails Of a particular Customer
CREATE OR REPLACE PROCEDURE training8am.customerservices(
    custid STRING
)
SQL SECURITY INVOKER
LANGUAGE SQL
AS
BEGIN
    SELECT *
    FROM training8am.services
    WHERE customerid = custid;
END;
```

```python
%sql
call training8am.customerservices("C103")
```
