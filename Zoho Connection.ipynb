```python
import json
import requests
from datetime import datetime, timedelta
from pyspark.sql import functions as F
from pyspark.sql.types import StructType, StructField, StringType

# =====================================================
# STEP 1 - GENERATE ACCESS TOKEN
# =====================================================
CLIENT_ID = "1000.V2WKQTUEOAJ240YEZYZ1TQ0PPM1RPX"
CLIENT_SECRET = "cecc832470fb7c453354f47559e4c2fb93142aad96"
REFRESH_TOKEN = "1000.a01ceeeec01c3c682c298afb8aa2f0b7.f94f2617baace2d345f1f5412ef2ccf4"

token_url = "https://accounts.zoho.in/oauth/v2/token"
payload = {
    "refresh_token": REFRESH_TOKEN,
    "client_id": CLIENT_ID,
    "client_secret": CLIENT_SECRET,
    "grant_type": "refresh_token"
}
token_response = requests.post(token_url, data=payload)
token_response.raise_for_status()
access_token = token_response.json()["access_token"]

# =====================================================
# STEP 2 - YESTERDAY 7 AM
# =====================================================
yesterday_7am = (
    datetime.now()
    .replace(hour=7, minute=0, second=0, microsecond=0)
    - timedelta(days=1)
)
criteria_date = (yesterday_7am + timedelta(hours=5, minutes=30)).strftime("%Y-%m-%dT%H:%M:%S+05:30")

# =====================================================
# STEP 3 - CALL COQL API
# =====================================================
url = "https://www.zohoapis.in/crm/v7/coql"
headers = {
    "Authorization": f"Zoho-oauthtoken {access_token}",
    "Content-Type": "application/json"
}

api_fields = [
    "id", "Student_ID", "First_Name", "Last_Name", "Email", "Phone", 
    "Owner", "Created_By", "Created_Time", "Registered_Date", 
    "Last_Activity_Time", "Modified_Time", "Course_Details", "TechStack", 
    "Total_Fees", "Fee_Paid", "Due_Amount", "Payment_Status", "Due_Date", 
    "Invoice_Type2", "GST_Number", "Source", "Location", 
    "gdriveextension__Drive_Folder_ID", "gdriveextension__Drive_URL", "Batch_ID_Comments"
]

query_payload = {
    "select_query": f"select {', '.join(api_fields)} from Contacts where Last_Activity_Time > '{criteria_date}'"
}

response = requests.post(url, headers=headers, json=query_payload)
response.raise_for_status()
response_payload = response.json()
api_data = response_payload.get("data", [])

if api_data:
    def to_string(value):
        return None if value is None else str(value)

    flattened_records = []

    for record in api_data:
        owner = record.get("Owner") or {}
        created_by = record.get("Created_By") or {}
        course_details = record.get("Course_Details")

        if isinstance(course_details, list):
            course_name = ", ".join(str(item) for item in course_details if item is not None)
        else:
            course_name = course_details

        flattened_records.append({
            "learner_record_id": to_string(record.get("id")),
            "learner_id": to_string(record.get("Student_ID")),
            "learner_name": to_string(record.get("First_Name") or record.get("Last_Name")),
            "last_name": to_string(record.get("Last_Name")),
            "email": to_string(record.get("Email")),
            "phone": to_string(record.get("Phone")),
            "learner_owner": to_string(owner.get("name")),
            "learner_owner_id": to_string(owner.get("id")),
            "created_by": to_string(created_by.get("name")),
            "created_by_id": to_string(created_by.get("id")),
            "created_time": to_string(record.get("Created_Time")),
            "registered_date": to_string(record.get("Registered_Date")),
            "last_activity_time": to_string(record.get("Last_Activity_Time")),
            "change_log_time": to_string(record.get("Modified_Time")),
            "course_name": to_string(course_name),
            "tech_stack": to_string(record.get("TechStack")),
            "total_fees": to_string(record.get("Total_Fees")),
            "fee_paid": to_string(record.get("Fee_Paid")),
            "due_amount": to_string(record.get("Due_Amount")),
            "payment_status": to_string(record.get("Payment_Status")),
            "due_date": to_string(record.get("Due_Date")),
            "invoice_type": to_string(record.get("Invoice_Type2")),
            "gst_number": to_string(record.get("GST_Number")),
            "source": to_string(record.get("Source")),
            "location": to_string(record.get("Location")),
            "drive_folder_id": to_string(record.get("gdriveextension__Drive_Folder_ID")),
            "drive_url": to_string(record.get("gdriveextension__Drive_URL")),
            "batch_id": to_string(record.get("Batch_ID_Comments")),
            "source_system": to_string(record.get("Source"))
        })

    flattened_schema = StructType([
        StructField("learner_record_id", StringType(), True),
        StructField("learner_id", StringType(), True),
        StructField("learner_name", StringType(), True),
        StructField("last_name", StringType(), True),
        StructField("email", StringType(), True),
        StructField("phone", StringType(), True),
        StructField("learner_owner", StringType(), True),
        StructField("learner_owner_id", StringType(), True),
        StructField("created_by", StringType(), True),
        StructField("created_by_id", StringType(), True),
        StructField("created_time", StringType(), True),
        StructField("registered_date", StringType(), True),
        StructField("last_activity_time", StringType(), True),
        StructField("change_log_time", StringType(), True),
        StructField("course_name", StringType(), True),
        StructField("tech_stack", StringType(), True),
        StructField("total_fees", StringType(), True),
        StructField("fee_paid", StringType(), True),
        StructField("due_amount", StringType(), True),
        StructField("payment_status", StringType(), True),
        StructField("due_date", StringType(), True),
        StructField("invoice_type", StringType(), True),
        StructField("gst_number", StringType(), True),
        StructField("source", StringType(), True),
        StructField("location", StringType(), True),
        StructField("drive_folder_id", StringType(), True),
        StructField("drive_url", StringType(), True),
        StructField("batch_id", StringType(), True),
        StructField("source_system", StringType(), True)
    ])

    zoho_df = (
        spark.createDataFrame(flattened_records, schema=flattened_schema)
        .select(
            F.trim(F.col("learner_record_id")).alias("learner_record_id"),
            F.trim(F.col("learner_id")).alias("learner_id"),
            F.trim(F.col("learner_name")).alias("learner_name"),
            F.trim(F.col("last_name")).alias("last_name"),
            F.trim(F.col("email")).alias("email"),
            F.trim(F.col("phone")).alias("phone"),
            F.trim(F.col("learner_owner")).alias("learner_owner"),
            F.trim(F.col("learner_owner_id")).alias("learner_owner_id"),
            F.trim(F.col("created_by")).alias("created_by"),
            F.trim(F.col("created_by_id")).alias("created_by_id"),
            F.trim(F.col("created_time")).alias("created_time"),
            F.trim(F.col("registered_date")).alias("registered_date"),
            F.trim(F.col("last_activity_time")).alias("last_activity_time"),
            F.trim(F.col("change_log_time")).alias("change_log_time"),
            F.trim(F.col("course_name")).alias("course_name"),
            F.trim(F.col("tech_stack")).alias("tech_stack"),
            F.col("total_fees").cast("decimal(12,2)").alias("total_fees"),
            F.col("fee_paid").cast("decimal(12,2)").alias("fee_paid"),
            F.col("due_amount").cast("decimal(12,2)").alias("due_amount"),
            F.trim(F.col("payment_status")).alias("payment_status"),
            F.to_date("due_date").alias("due_date"),
            F.trim(F.col("invoice_type")).alias("invoice_type"),
            F.trim(F.col("gst_number")).alias("gst_number"),
            F.trim(F.col("source")).alias("source"),
            F.trim(F.col("location")).alias("location"),
            F.trim(F.col("drive_folder_id")).alias("drive_folder_id"),
            F.trim(F.col("drive_url")).alias("drive_url"),
            F.current_timestamp().alias("ingestion_timestamp"),
            F.trim(F.col("batch_id")).alias("batch_id"),
            F.trim(F.col("source_system")).alias("source_system")
        )
    )

    zoho_df.createOrReplaceTempView("zoho_api_staging_updates")
    zoho_df.show()
    
    # Run the merge inside the conditional block to prevent running it when there is no data
    spark.sql("""
        MERGE INTO databrickstraining.bronze.leanrers_bronze AS target
        USING (
            SELECT 
                learner_record_id,
                learner_id,
                learner_name,
                last_name,
                email,
                phone,
                learner_owner,
                learner_owner_id,
                created_by,
                created_by_id,
                CAST(created_time AS TIMESTAMP) AS created_time,
                registered_date,
                last_activity_time,
                change_log_time,
                course_name,
                tech_stack,
                total_fees,
                fee_paid,
                due_amount,
                payment_status,
                CAST(due_date AS STRING) AS due_date,
                invoice_type,
                gst_number,
                source,
                location,
                drive_folder_id,
                drive_url,
                batch_id,
                source_system
            FROM zoho_api_staging_updates
        ) AS source
        ON target.learner_record_id = source.learner_record_id
        WHEN MATCHED THEN
            UPDATE SET 
                target.learner_id = source.learner_id,
                target.learner_name = source.learner_name,
                target.last_name = source.last_name,
                target.email = source.email,
                target.phone = source.phone,
                target.learner_owner = source.learner_owner,
                target.learner_owner_id = source.learner_owner_id,
                target.created_by = source.created_by,
                target.created_by_id = source.created_by_id,
                target.created_time = source.created_time,
                target.registered_date = source.registered_date,
                target.last_activity_time = source.last_activity_time,
                target.change_log_time = source.change_log_time,
                target.course_name = source.course_name,
                target.tech_stack = source.tech_stack,
                target.total_fees = source.total_fees,
                target.fee_paid = source.fee_paid,
                target.due_amount = source.due_amount,
                target.payment_status = source.payment_status,
                target.due_date = source.due_date,
                target.invoice_type = source.invoice_type,
                target.gst_number = source.gst_number,
                target.source = source.source,
                target.location = source.location,
                target.drive_folder_id = source.drive_folder_id,
                target.drive_url = source.drive_url,
                target.batch_id = source.batch_id,
                target.source_system = source.source_system
        WHEN NOT MATCHED THEN
            INSERT (
                learner_record_id, learner_id, learner_name, last_name, email, phone, 
                learner_owner, learner_owner_id, created_by, created_by_id, created_time, 
                registered_date, last_activity_time, change_log_time, course_name, tech_stack, 
                total_fees, fee_paid, due_amount, payment_status, due_date, invoice_type, 
                gst_number, source, location, drive_folder_id, drive_url, batch_id, source_system
            )
            VALUES (
                source.learner_record_id, source.learner_id, source.learner_name, source.last_name, source.email, source.phone, 
                source.learner_owner, source.learner_owner_id, source.created_by, source.created_by_id, source.created_time, 
                source.registered_date, source.last_activity_time, source.change_log_time, source.course_name, source.tech_stack, 
                source.total_fees, source.fee_paid, source.due_amount, source.payment_status, source.due_date, source.invoice_type, 
                source.gst_number, source.source, source.location, source.drive_folder_id, source.drive_url, source.batch_id, source.source_system
            )
    """)
else:
    print(response_payload)
    print("No records returned from Zoho.")
```
