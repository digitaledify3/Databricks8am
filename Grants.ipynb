Grants - Select,Modify,Use Schema, Execute,Create Table

Schema Level -
- Create Table - This create table grant will allow the user to create tables under that particular schema

```python
%sql
show schemas in workspace
```

```python
%sql
GRANT CREATE TABLE ON SCHEMA workspace.training8am to `account users`
```

- Since we are not having a catalog
When we say workspace.training8am, Databricks assume the catalog name as workspace

It will be searching for a workspace schema under the workspace catalog

- Modify on the Schema-
- Modify is applicable to a table
- Modify allows the user to INSERT,UPDATE,DELETE,TRUNCATE in a table

- If you give modify to the schema, the tables under that schema will inhert this grant.

```python
%sql
GRANT MODIFY ON SCHEMA workspace.training8am to `kiran@digitaledify.ai`

-- The user kiran@digitaledify.ai will be getting access to modify all the tables that are present under schema training8am.
```

Select
- The uses will be getting access to read the inside a particualar table

Select on Schema Level
- ALl the tables under that particular schmea will be given read access to the service principle

```python
%sql
GRANT SELECT ON  workspace.training8am to 'kiran@digitaledify.ai';


-- Kiran will be getting access to read all the tables under training8am
```

USE SCHEMA
- It means you are giving permission to enter or access a schema

- The user has
GRANT SELECT ON workspace.training8am to 'kiran@digitaledify.ai'

- But user dont have
USE SCHEMA grant

Will the user be able to query the tables under training8am?
The use cannot read the table


```python
%sql
GRANT USE SCHEMA on SCHEMA workspace.training8am to 'kiran@digitaledify.ai'

-- Kiran is having access to the schema. Now create , select, modify will work under this schema
```
