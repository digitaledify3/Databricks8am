```python
%sql
insert into databrickstraining.bronze.leanrers_bronze
(
    learner_record_id,
    learner_id,
    learner_name,
    last_name,
    email,
    phone,

    learner_owner,
    learner_owner_id,

    created_by,
    created_by_id,

    created_time,
    registered_date,
    last_activity_time,
    change_log_time,
    course_name,
    tech_stack,

    total_fees,
    fee_paid,
    due_amount,
    payment_status,
    due_date,

    invoice_type,
    gst_number,

    source,
    location,

    drive_folder_id,
    drive_url,

    ingestion_timestamp,
    batch_id,
    source_system
)
SELECT
TRIM(CAST(`Record Id` AS STRING)) AS learner_record_id,
TRIM(CAST(`DL Learner ID` AS STRING)) AS learner_id,
TRIM(CAST(`Learner Name` AS STRING)) AS learner_name,
TRIM(CAST(`Last Name` AS STRING)) AS last_name,
TRIM(CAST(`Email` AS STRING)) AS email,
TRIM(CAST(`Phone` AS STRING)) AS phone,

TRIM(CAST(`Learner Owner` AS STRING)) AS learner_owner,
TRIM(CAST(`Learner Owner.id` AS STRING)) AS learner_owner_id,

TRIM(CAST(`Created By` AS STRING)) AS created_by,
TRIM(CAST(`Created By.id` AS STRING)) AS created_by_id,

CAST(`Created Time` AS timestamp) AS created_time,
CAST(`Registered Date` AS String) AS registered_date,
CAST(`Last Activity Time` AS String) AS last_activity_time,
CAST(`Change Log Time` AS String) AS change_log_time,
TRIM(CAST(`Course Details` AS STRING)) AS course_name,
TRIM(CAST(`TechStack` AS STRING)) AS tech_stack,

CAST(`Total - Fees` AS DECIMAL(12,2)) AS total_fees,
CAST(`Fee - Paid` AS DECIMAL(12,2)) AS fee_paid,
CAST(`Due-Amount` AS DECIMAL(12,2)) AS due_amount,
TRIM(CAST(`Payment Status` AS STRING)) AS payment_status,
CAST(`Due - Date` AS DATE) AS due_date,

TRIM(CAST(`Invoice Type` AS STRING)) AS invoice_type,
TRIM(CAST(`GST Number` AS STRING)) AS gst_number,

TRIM(CAST(`Source` AS STRING)) AS source,
TRIM(CAST(`Location` AS STRING)) AS location,

TRIM(CAST(`Drive Folder ID` AS STRING)) AS drive_folder_id,
TRIM(CAST(`Drive URL` AS STRING)) AS drive_url,

CURRENT_TIMESTAMP() AS ingestion_timestamp,
TRIM(CAST(`Batch ID Comments` AS STRING)) AS batch_id,
TRIM(CAST(`Source` AS STRING)) AS source_system
FROM (select * from learnerhistory where `Created Time` != 'Fullstack')


```
